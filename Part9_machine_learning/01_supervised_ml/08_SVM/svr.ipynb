## SVM- Regression

```python
import pandas as pandas
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score

from sklearn.svm import SVR

from sklearn import datasets
```

```python
df = datasets.load_diabetes(as_frame=True).frame

df.head()
```

```python
df.info()
```

```python
X = df.drop('target', axis=1)
y = df['target']

X.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3, random_state=42)
X_train.head()
```

- input features looks scaled
- but target var is not scaled

```python
y_scaler = StandardScaler()
y_train_scaled = y_scaler.fit_transform(y_train)
y_test_scaled = y_scaler.transform(y_test)
```

```python
y_train.values.reshape(-1,1)
```

```python
y_scaler = StandardScaler()
y_train_scaled = y_scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
y_test_scaled = y_scaler.transform(y_test.values.reshape(-1,1)).ravel()
```

```python
y_test_scaled
```

```python
model = SVR()
model.fit(X_train,y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("Training - r2 Score: ",r2_score( y_train_scaled,y_train_pred_scaled)) # for only r2 score r2_score(y_pred_scaled, y_test_scaled) eqv to r2_score(y_pred, y_test) but in case of MSE,MAE,etc we need to conver back to original data
print("Testing - r2 Score: ",r2_score( y_test_scaled,y_test_pred_scaled))
```

- above we see slight overfitting as score of testing is less than training

```python
model = SVR(kernel='linear')
model.fit(X_train,y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("Training - r2 Score: ",r2_score( y_train_scaled,y_train_pred_scaled))
print("Testing - r2 Score: ",r2_score( y_test_scaled,y_test_pred_scaled))
```

- Above is less overfit

```python
model = SVR(kernel='poly')
model.fit(X_train,y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("Training - r2 Score: ",r2_score( y_train_scaled,y_train_pred_scaled))
print("Testing - r2 Score: ",r2_score( y_test_scaled,y_test_pred_scaled))
```

- Above large overfitting case

```python
model = SVR(kernel='sigmoid')
model.fit(X_train,y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("Training - r2 Score: ",r2_score( y_train_scaled,y_train_pred_scaled))
print("Testing - r2 Score: ",r2_score( y_test_scaled,y_test_pred_scaled))
```

- Abve is bad case

### HyperParameter Tuning using GridSearchCV

```python
param_grid = {
    'kernel':('linear', 'rbf','poly',),
    'C':[1, 2, 5, 10, 50, 100],
    'epsilon':[0.01,0.1,0.2,0.3,0.5]
}

from sklearn.model_selection import GridSearchCV

# grid = GridSearchCV(
#     estimato=rmodel,
#     param_grid=param_grid,
#     cv=5,
#     scoring="r2_score"
# )
grid = GridSearchCV(
    model,
    param_grid,
    cv=5,
    scoring="r2"
)

grid.fit(X_train,y_train_scaled)

y_test_pred_scaled = grid.predict(X_test)
y_train_pred_scaled = grid.predict(X_train)

print("Training - r2 Score: ",r2_score( y_train_scaled,y_train_pred_scaled))
print("Testing - r2 Score: ",r2_score( y_test_scaled,y_test_pred_scaled))
```

```python
print("best parameters: ", grid.best_params_)
```

```python
model = grid.best_estimator_
model.fit(X_train,y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("Training - r2 Score: ",r2_score( y_train_scaled,y_train_pred_scaled))
print("Testing - r2 Score: ",r2_score( y_test_scaled,y_test_pred_scaled))
```

LinearSVR- SVR(kernel='linear')

```python
from sklearn.svm import LinearSVR

model2 = LinearSVR(C=10,epsilon=0.1)

model2.fit(X_train,y_train_scaled)

y_test_pred_scaled = model2.predict(X_test)
y_train_pred_scaled = model2.predict(X_train)

print("Training - r2 Score: ",r2_score( y_train_scaled,y_train_pred_scaled))
print("Testing - r2 Score: ",r2_score( y_test_scaled,y_test_pred_scaled))
```

```python
from sklearn.svm import LinearSVR

model2 = LinearSVR(C=10,epsilon=0.1,max_iter=10000)

model2.fit(X_train,y_train_scaled)

y_test_pred_scaled = model2.predict(X_test)
y_train_pred_scaled = model2.predict(X_train)

print("Training - r2 Score: ",r2_score( y_train_scaled,y_train_pred_scaled))
print("Testing - r2 Score: ",r2_score( y_test_scaled,y_test_pred_scaled))
```
